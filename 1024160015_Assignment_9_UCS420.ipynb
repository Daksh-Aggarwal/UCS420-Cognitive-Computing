```python
# Q1
import cv2
import easyocr
import matplotlib.pyplot as plt
import numpy as np

reader = easyocr.Reader(['en'], gpu=False)
img = cv2.imread('medicine.jpeg')
res = reader.readtext('medicine.jpeg')

for b, text, prob in res:
    if prob > 0.3:
        print(f"{text} ({prob:.2f})")
        pts = np.array(b, dtype=np.int32)
        cv2.polylines(img, [pts], True, (0, 255, 0), 2)

plt.figure(figsize=(10, 10))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
# Q2
from deepface import DeepFace
import cv2
import matplotlib.pyplot as plt

try:
    analysis = DeepFace.analyze(img_path="don_pollo.jpeg", actions=['emotion'])
    print(analysis[0]['dominant_emotion'])
    print(analysis[0]['emotion'])
except Exception as e:
    print(e)

face_img = cv2.imread('don_pollo.jpeg')
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(face_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
