```python
# Q2
# a
arr = np.array([1, 2, 3, 4, 5])
arr = arr[::-1]
print(arr)

# b
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

vals, count = np.unique(x, return_counts=True)
# print(vals)
# print(count)

mode = vals[np.argmax(count)]
indices = np.flatnonzero(x == mode)
print(f"x's mode: {mode}")
print(f"x's corresponding indices: {indices}")

vals, count = np.unique(y, return_counts=True)
# print(vals)
# print(count)

mode = vals[np.argmax(count)] # found the first most frequent value, both 1 and 3 were possible
indices = np.flatnonzero(y == mode)
print(f"y's mode: {mode}")
print(f"y's corresponding indices: {indices}")
```

```python
# Q4
daksh = np.linspace(10, 100, 25)
print("\nDimensions:", np.ndim(daksh))
print("Shape:", np.shape(daksh))
print("Total elements:", np.size(daksh))
print("Data type of each element:", daksh.dtype)
print("Total bytes consumed:", daksh.nbytes)

transpose_reshape = daksh.reshape(1, -1)
print(f"\nTranspose using reshape(): \n{transpose_reshape}\n")

transpose_T = daksh.T
print(f"Using T attribute: \n{transpose_T}")
```

```python
# Q5
ucs420_daksh = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35])
ucs420_daksh = np.reshape(ucs420_daksh, (3, 4))
print(ucs420_daksh)

print(f"\nMean: {ucs420_daksh.mean()}")
print(f"Median: {np.median(ucs420_daksh)}")
print(f"Max: {ucs420_daksh.max()}")
print(f"Min: {ucs420_daksh.min()}")
print(f"Unique elements: {np.unique(ucs420_daksh)}\n")

reshaped_ucs420_daksh = np.reshape(ucs420_daksh, (4, 3))
print(f"Reshaped array: \n{reshaped_ucs420_daksh}\n")

resized_ucs420_daksh = np.resize(ucs420_daksh, (2, 3))
print(f"Resized array: \n{resized_ucs420_daksh}")
```
