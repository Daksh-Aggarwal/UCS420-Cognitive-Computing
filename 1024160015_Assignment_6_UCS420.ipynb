```python
# Q3
# a
original = np.array([1, 2, 3, 4, 5, 6])

# b
subset = original[1:5]
print(subset)

# c
subset[0] = 999
print(original)
print(subset)

# d
subset = original[1:5].copy()
subset[0] = 500
print(original)
print(subset)

# e
arr = np.arange(1, 13)
print(arr)
arr = np.reshape(arr, (3, 4))
print(arr)

# f
print(f"First row: {arr[0, :]}")
print(f"Last row: {arr[2, :]}")
print(f"Second column: {arr[:, 1]}")
print(f"Last column: {arr[:, 3]}")
print(arr[0:1, 1:3])

# g
arr_flattened = arr.flatten()
arr_raveled = arr.ravel()

# h
arr_raveled[0] = 67
print(arr)
print(arr_raveled)

# i
arr_flattened[1] = 500
print(arr)
print(arr_flattened)

# j
print(f"Shape: {original.shape}")
print(f"Dimensions: {original.ndim}")
print(f"Size: {np.size(original)}")
print(f"dtype: {original.dtype}")
```

```python
# Q4
X = np.array([
[6, 70, 3],
[5, 50, 6],
[8, 80, 2],
[4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# a
print("Shape:", X.shape)
print("Dimensions:", X.ndim)

# b
X_transpose = X.T
print(f"\nX.T: \n{X_transpose}")

# c
XTX = X.T @ X
print(f"\nX.T @ X: \n{XTX}")

# d
inverse_XTX = np.linalg.inv(XTX)
print(f"\nInverse of X.T @ X: \n{inverse_XTX}")

# e
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print(f"\nCoefficients: {beta}")

# f
# beta[0] -> coefficient for Feature 1
# beta[1] -> coefficient for Feature 2
# beta[2] -> coefficient for Feature 3

# g
new_user = np.array([5, 40, 7])

predicted_assistance_score = new_user @ beta
print("\nPredicted assistance score:", predicted_assistance_score)
```
